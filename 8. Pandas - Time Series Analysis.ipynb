```python
l = pd.read_csv('TSA Dataset\\login 1.csv',header=None)
l = l[[1,2]]
l
```

```python
l.rename(columns={1:'user_id',2:'login_time'},inplace=True)
```

```python
l
```

```python
l['login_time'] = pd.to_datetime(l['login_time'])
```

```python
l['user_id'].value_counts()
```

```python
# avg visit time for 413 user
user_df = l[l['user_id']==413]
user_df
```

```python
user_df['shifted']=user_df['login_time'].shift(1)
```

```python
user_df
```

```python
user_df['login_time'] - user_df['shifted']
```

```python
(user_df['login_time'] - user_df['shifted']).mean()
```
